# ACN-Data session EDA

MVP ingest uses `DataClient`, not ACN-Sim. Canonical snapshot:

```text
uv run chargeopt data pull
```

This notebook is exploration only. Token from `CHARGEOPT_ACN_TOKEN`, default `DEMO_TOKEN`.


````python
from datetime import datetime
from zoneinfo import ZoneInfo

from acnportal.acndata import DataClient

from chargeopt.config import load_config
from chargeopt.data.acn import normalize_session
````

````python
config = load_config()
````

````python
tz = ZoneInfo(config.data.timezone)
````

````python
start = config.data.start.replace(tzinfo=tz) if config.data.start.tzinfo is None else config.data.start
````

````python
end = start.replace(day=start.day)  # same calendar day peek; widen via config for full pull
end = start + __import__("datetime").timedelta(days=1)
````

````python
client = DataClient("DEMO_TOKEN")
raw = next(client.get_sessions_by_time(config.data.site, start=start, end=end))
print(sorted(raw.keys()))
print(normalize_session(raw, site=config.data.site))
````
